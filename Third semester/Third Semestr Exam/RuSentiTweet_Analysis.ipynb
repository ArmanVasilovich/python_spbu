```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_df_path = '/content/drive/MyDrive/Colab Notebooks/Third semester/Fine-Tuning/RuBert-Data/rusentitweet_train.csv'
test_df_path = '/content/drive/MyDrive/Colab Notebooks/Third semester/Fine-Tuning/RuBert-Data/rusentitweet_test.csv'
```

```python
train_df = pd.read_csv(train_df_path)
test_df = pd.read_csv(test_df_path)
```

```python
sns.set(style="whitegrid")
target_col = 'label' if 'label' in train_df.columns else train_df.columns[1]

ax = sns.countplot(data=train_df, x=target_col, palette='viridis')
plt.title('Распределение классов (Class Distribution)', fontsize=14)
plt.xlabel('Класс тональности')
plt.ylabel('Количество примеров')
for container in ax.containers:
    ax.bar_label(container)
```

```python
train_df['length'] = train_df[train_df.columns[0]].apply(lambda x: len(str(x).split()))
sns.histplot(train_df['length'], bins=30, kde=True, color='teal')
plt.title('Распределение длины твитов (слов)', fontsize=14)
plt.xlabel('Количество слов')
plt.ylabel('Частота')
```

```python
sns.boxplot(data=train_df, x=target_col, y='length', palette='viridis')
plt.title('Длина текста в зависимости от класса', fontsize=14)
plt.ylim(0, 30)

plt.tight_layout()
plt.show()
```

```python
print(f"Всего примеров в обучающей выборке: {len(train_df)}")
print(f"Средняя длина твита: {train_df['length'].mean():.1f} слов")
print(f"Максимальная длина: {train_df['length'].max()} слов")
```
