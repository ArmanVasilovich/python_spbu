```python
x,b = symbols('x,b')
f = 0.5*x+b
f
```

```python
y1 = f.subs(x,2)
y2 = f.subs(x,1)
y3 = f.subs(x,3)
```

```python
MSE = ((1-y1)**2+ (3-y2)**2 + (4-y3)**2)/3
MSE
```

```python
MSE = simplify(MSE)
MSE
```

```python
d = diff(MSE,b)
d
```

```python
result = solve(d)
result
```

```python
f_new = lambda x: 0.5*x+result
```
